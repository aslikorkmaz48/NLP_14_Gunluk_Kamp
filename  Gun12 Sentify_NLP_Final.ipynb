```python
# 1. Gerekli kütüphaneleri kur (Eğer kurulu değilse)
!pip install fpdf requests beautifulsoup4 transformers matplotlib

import requests
from bs4 import BeautifulSoup
import pandas as pd
from transformers import pipeline
import matplotlib.pyplot as plt
from fpdf import FPDF
import datetime
from google.colab import files
import os

def nlp_developer_final_task(konu="Elon Musk"):
    print(f"🔍 '{konu}' konusu üzerine NLP Analizi başlatılıyor...")

    # --- ADIM 1: WEB SCRAPING (Veri Toplama) ---
    url = f"https://news.google.com/rss/search?q={konu}&hl=tr&gl=TR&ceid=TR:tr"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, features="xml")
    items = soup.find_all("item")[:10] # Son 10 haber

    # --- ADIM 2: NLP SENTIMENT ANALYSIS (BERT Model) ---
    print("🤖 BERT Modeli yükleniyor ve analiz yapılıyor...")
    # Türkçe için optimize edilmiş BERT modelini kullanıyoruz
    sentiment_pipeline = pipeline("sentiment-analysis", model="savasy/bert-base-turkish-sentiment-cased")

    results = []
    for item in items:
        title = item.title.text
        prediction = sentiment_pipeline(title)[0]
        results.append({
            "Haber": title,
            "Duygu": prediction['label'],
            "Skor": f"%{int(prediction['score']*100)}"
        })
    df = pd.DataFrame(results)

    # --- ADIM 3: DATA VISUALIZATION (Görselleştirme) ---
    counts = df['Duygu'].value_counts()
    plt.figure(figsize=(6, 4))
    plt.pie(counts, labels=counts.index, autopct='%1.1f%%', colors=['#2ecc71', '#e74c3c', '#3498db'])
    plt.title(f"{konu} - Duygu Analizi Dağılımı")
    plt.savefig("analiz_grafik.png")
    plt.close()

    # --- ADIM 4: PROFESSIONAL PDF REPORTING (Raporlama) ---
    pdf = FPDF()
    pdf.add_page()

    # Başlık ve Tarih
    pdf.set_font("Arial", 'B', 16)
    pdf.cell(200, 10, txt="NLP Developer Intelligence Report", ln=True, align='C')
    pdf.set_font("Arial", size=10)
    pdf.cell(200, 10, txt=f"Analysis: {konu} | Date: {datetime.datetime.now().strftime('%d/%m/%Y')}", ln=True, align='C')

    # Grafiği Ekle
    pdf.image("analiz_grafik.png", x=50, y=35, w=110)
    pdf.ln(90) # Grafikten sonra boşluk bırak

    # Haber Listesi
    pdf.set_font("Arial", 'B', 12)
    pdf.cell(0, 10, txt="Analyzed News Feed:", ln=True)
    pdf.set_font("Arial", size=9)

    for _, row in df.iterrows():
        # Türkçe karakterleri ASCII'ye temizleyerek PDF hatasını önle
        clean_text = row['Haber'].encode('ascii', 'ignore').decode('ascii')
        msg = f"- {clean_text[:85]}... [{row['Duygu']} {row['Skor']}]"
        pdf.multi_cell(0, 8, txt=msg)

    file_name = f"{konu.replace(' ', '_')}_Final_Report.pdf"
    pdf.output(file_name)

    print(f"✅ Rapor Hazır: {file_name}")
    files.download(file_name) # Otomatik indirme

# SİSTEMİ ÇALIŞTIR
nlp_developer_final_task("Teknoloji")
```
